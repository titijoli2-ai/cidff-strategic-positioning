# Axis 3 — Funding and Financial Sustainability

## Research Question

How can funding opportunities and financial constraints
influence the strategic positioning of CIDFF Rhône–Arc Alpin?

## Objectives

- Identify available public funding data.
- Explore funding differences across the four departments:
  Isère, Rhône, Savoie and Haute-Savoie.
- Examine possible financial risks and opportunities.
- Support strategic decisions using available evidence.

## Methodology

This analysis will use public datasets and official sources.

Missing financial information will not be estimated.
Funding amounts will only be compared when their
definitions and periods are compatible.

## Data scope

The analysis uses State payment records for 2020–2024, published
in the association funding annexes to the 2022–2026 Finance Bills.

It covers beneficiaries located in Isère, Rhône, Savoie and Haute-Savoie.
Their location does not necessarily show where funded activities take place.

Recorded payments include subsidies, service payments and other flows.
They do not represent associations' total income or all public funding.

Amounts are in current euros. The years 2020–2021 include the pandemic
period and should be interpreted with caution.

```python
# Keep shared analysis fields and essential quality flags

# No geographic field-shift correction was applied to these years
for year in [2020, 2021, 2022, 2024]:
    territorial_funding[year]["geography_corrected"] = False
analysis_columns = [
    "year",
    "department_code",
    "municipality_code",
    "municipality_name",
    "siren",
    "nic",
    "association_name",
    "budget_program",
    "amount",
    "funding_purpose",
    "agreement",
    "siren_original",
    "siren_corrected",
    "siren_format_valid",
    "geography_corrected",
    "identical_record_flag"
]

# Combine the five years without removing any records
funding = pd.concat(
    [df[analysis_columns] for df in territorial_funding.values()],
    ignore_index=True
)

funding.shape
```

## Data quality decisions

- Fully empty rows were removed.
- Amounts were converted to numbers. One unknown amount in the national
  2020 file was kept missing; it falls outside the study departments.
- Leading zeros were restored in four-digit municipality codes.
- In 2023, 359 records were corrected using a detected field shift.
  This was a structural correction, not an external validation.
  Another 89 records remained geographically unresolved and could
  not be included in the territorial selection.
- Two incomplete SIRENs were corrected using matching records from
  later years. Two records for PASS LAS 21 still have no SIREN or NIC.
- Identical records were flagged and retained because the available
  fields do not establish whether they are duplicate payments.
- One zero-amount record was retained.
- Historical purposes were stored separately and were not treated
  as confirmed purposes of 2024 payments.

```python
# Treat missing, blank and zero-only purposes as unavailable
purpose = funding["funding_purpose"].str.strip()

unavailable_purpose = (
    purpose.isna()
    | purpose.isin(["", "0", "0.0"])
)

funding.assign(
    unavailable_purpose=unavailable_purpose
).groupby("year").agg(
    record_count=("amount", "size"),
    unavailable_purpose_count=("unavailable_purpose", "sum"),
    unavailable_purpose_share=("unavailable_purpose", "mean")
)
```

```python
# Find historical records with an available funding purpose
historical = funding.loc[
    funding["year"].isin([2022, 2023])
    & funding["siret"].notna()
    & funding["funding_purpose"].notna()
    & ~funding["funding_purpose"].str.strip().isin(["", "0", "0.0"]),
    ["siret", "budget_program"]
].drop_duplicates()

# Select 2024 records with an unavailable purpose
current = funding.loc[
    funding["year"].eq(2024)
    & (
        funding["funding_purpose"].isna()
        | funding["funding_purpose"].str.strip().isin(["", "0", "0.0"])
    ),
    ["siret", "budget_program"]
]

# Check matches without copying historical purposes
coverage = current.merge(
    historical,
    on=["siret", "budget_program"],
    how="left",
    indicator=True,
    validate="many_to_one"
)

print("2024 records without a purpose:", len(coverage))
print("Records with a historical match:", coverage["_merge"].eq("both").sum())
print(f"Match share: {coverage['_merge'].eq('both').mean():.1%}")
```

```python
# Count distinct historical purposes for each establishment and program
historical_purposes = funding.loc[
    funding["year"].isin([2022, 2023])
    & funding["siret"].notna()
    & funding["funding_purpose"].notna()
    & ~funding["funding_purpose"].str.strip().isin(["", "0", "0.0"]),
    ["siret", "budget_program", "funding_purpose"]
]

purpose_counts = (
    historical_purposes.groupby(["siret", "budget_program"])
    ["funding_purpose"].nunique()
    .rename("historical_purpose_count")
    .reset_index()
)

purpose_check = current.merge(
    purpose_counts,
    on=["siret", "budget_program"],
    how="left",
    validate="many_to_one"
)

print("One historical purpose:",
      purpose_check["historical_purpose_count"].eq(1).sum())
print("Several historical purposes:",
      purpose_check["historical_purpose_count"].gt(1).sum())
print("No historical match:",
      purpose_check["historical_purpose_count"].isna().sum())
```

```python
# Keep only establishment-program pairs with one historical purpose
unique_pairs = purpose_counts.loc[
    purpose_counts["historical_purpose_count"].eq(1),
    ["siret", "budget_program"]
]

historical_context = (
    historical_purposes
    .merge(unique_pairs, on=["siret", "budget_program"])
    .drop_duplicates(["siret", "budget_program"])
    .rename(columns={"funding_purpose": "historical_purpose_context"})
)

# Add historical context without changing current funding purposes
funding = funding.merge(
    historical_context,
    on=["siret", "budget_program"],
    how="left",
    validate="many_to_one"
)

# Keep this context only for 2024 records without a purpose
needs_context = funding["year"].eq(2024) & (
    funding["funding_purpose"].isna()
    | funding["funding_purpose"].str.strip().isin(["", "0", "0.0"])
)

funding.loc[
    ~needs_context, "historical_purpose_context"
] = pd.NA

print("Records with historical context:",
      funding["historical_purpose_context"].notna().sum())
```

### What share of CIDFF's recorded State payments comes from program 137?

```python
# Calculate total State payments to CIDFF by year
cidff_total = cidff_raa.groupby("year")["amount"].sum()

# Calculate program 137 payments to CIDFF by year
cidff_137 = cidff_raa[
    cidff_raa["budget_program"] == "137"
].groupby("year")["amount"].sum()

# Compare program 137 with total recorded State payments
cidff_funding_mix = pd.DataFrame({
    "total_state_payments": cidff_total,
    "program_137_payments": cidff_137
})

cidff_funding_mix["program_137_share_pct"] = (
    cidff_137 / cidff_total * 100
)

cidff_funding_mix.round(2)
```

```python
# Prepare the two shares for each year
shares = pd.DataFrame({
    "Program 137": cidff_funding_mix["program_137_share_pct"]
})

shares["Other State programs"] = 100 - shares["Program 137"]

# Plot the composition of recorded State payments
shares.plot(
    kind="bar",
    stacked=True,
    figsize=(8, 5),
    color=["#7851A9", "#D5D8DC"]
)

plt.title("CIDFF — Composition of recorded State payments")
plt.xlabel("Year")
plt.ylabel("Share of annual recorded State payments (%)")
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.legend(title="", loc="upper left", bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```
